### 🧠 File Access Modes Table:
| Mode | Meaning | Creates File? | Truncates Existing? |
| :--- | :--- | :--- | :--- |
| `'r'` | Read text (Default) | ❌ Error if missing | ❌ No |
| `'w'` | Write text | ✅ Yes | ⚠️ Yes (Overwrites) |
| `'a'` | Append text at EOF | ✅ Yes | ❌ No (Appends) |
| `'r+'` | Read and Write | ❌ Error if missing | ❌ No |
| `'rb'` / `'wb'` | Read/Write raw Binary | Mode dependent | Depends on `w`/`r` |


### 1. Writing to a Text File with Context Managers (`with open`)
The `with` statement guarantees that the file stream is cleanly flushed and closed when execution leaves the block, even if an unhandled exception occurs.


```python
import os

sample_filename = "training_metrics.csv"

# Writing header and initial training rows
with open(sample_filename, "w", encoding="utf-8") as f:
    f.write("epoch,train_loss,val_loss,val_accuracy\n")
    f.write("1,0.6842,0.6120,0.7250\n")
    f.write("2,0.4510,0.3980,0.8420\n")

print(f"File '{sample_filename}' created successfully. File size: {os.path.getsize(sample_filename)} bytes")
```

### 2. Appending New Data to an Existing File (`'a'` mode)
Append mode adds new lines at the end of the file without deleting existing content.


```python
# Appending epoch 3 metrics
with open(sample_filename, "a", encoding="utf-8") as f:
    f.write("3,0.3120,0.2850,0.9140\n")
    f.write("4,0.1980,0.1820,0.9560\n")

print(f"Appended epoch 3 & 4 data to '{sample_filename}'.")
```

### 3. Reading Files: `read()`, `readline()`, and `readlines()`
- `f.read()`: Reads the entire file into a single string.
- `f.readline()`: Reads one line at a time as a string.
- `f.readlines()`: Reads all lines into a list of strings.


```python
print("--- 1. Iterating line-by-line (Memory-Efficient) ---")
with open(sample_filename, "r", encoding="utf-8") as f:
    for line_num, line in enumerate(f, start=1):
        print(f"Line {line_num}: {line.strip()}")

print("\n--- 2. Reading all lines with readlines() ---")
with open(sample_filename, "r", encoding="utf-8") as f:
    all_lines = f.readlines()
    print("Total rows:", len(all_lines))
    print("Header:", all_lines[0].strip())
    print("Last row:", all_lines[-1].strip())
```

### 4. File Pointer Navigation: `tell()` and `seek()`
- `f.tell()`: Returns current byte offset position in the file.
- `f.seek(offset, whence)`: Moves the file pointer cursor to a specific byte location (`0` = beginning of file).


```python
with open(sample_filename, "r", encoding="utf-8") as f:
    print("Initial file cursor position (tell):", f.tell())
    
    header = f.readline()
    print(f"Header read: {header.strip()} | Cursor position: {f.tell()}")
    
    # Rewind pointer back to beginning of file
    f.seek(0)
    print("Cursor position after f.seek(0):", f.tell())
    
    first_char = f.read(5)
    print("First 5 characters:", repr(first_char))
```

### 5. OS & Pathlib: Directory Creation and Scanning
Modern Python uses `pathlib.Path` for cross-platform file paths and automated folder structures.


```python
from pathlib import Path

# Create nested directory tree for AI checkpoints
checkpoint_dir = Path("ai_experiment_logs/run_01")
checkpoint_dir.mkdir(parents=True, exist_ok=True)
print("Created checkpoint directory:", checkpoint_dir.resolve())

# Write config JSON inside the folder
config_file = checkpoint_dir / "experiment_config.txt"
with open(config_file, "w", encoding="utf-8") as f:
    f.write("Model: Transformer-Encoder\nDataset: CIFAR-100\nOptimizer: AdamW\n")

# Scan directory contents
print("\nFiles in directory:")
for file_path in checkpoint_dir.iterdir():
    print(f"  Found file: {file_path.name} ({file_path.stat().st_size} bytes)")
```

### 🏋️ Student Practice Challenge:
Write a Python script that reads `"training_metrics.csv"` generated above, extracts the `val_accuracy` values from each row, calculates the average validation accuracy, and prints the result.


```python
# Write your solution below:
with open("training_metrics.csv", "r", encoding="utf-8") as f:
    lines = f.readlines()
    header = lines[0].strip().split(",")
    val_acc_idx = header.index("val_accuracy")
    
    accuracies = []
    for line in lines[1:]:
        parts = line.strip().split(",")
        accuracies.append(float(parts[val_acc_idx]))

avg_acc = sum(accuracies) / len(accuracies)
print(f"Validation Accuracies: {accuracies}")
print(f"Mean Validation Accuracy: {avg_acc:.4f} ({avg_acc * 100:.2f}%)")
```
