### 🧠 Theory & Concepts:
- **Class:** A user-defined prototype / blueprint for an object that defines a set of attributes and methods.
- **Object:** An instance of a class with concrete state values.
- **`self`:** A reference to the current instance of the class passed as the first parameter to instance methods.
- **`__init__(self, ...)`:** The constructor method called automatically when an object is instantiated.


### 1. Basic Class Definition & Class Attributes
Let's define a simple `Snake` and `AIModel` class to understand attributes and methods.


```python
# Basic Class Definition
class Snake:
    # Class attribute (shared by all instances)
    species = "Reptile"
    
    def __init__(self, name: str, venomous: bool = False):
        # Instance attributes (unique to this instance)
        self.name = name
        self.venomous = venomous
        
    def change_name(self, new_name: str):
        """Instance method modifying object state."""
        old_name = self.name
        self.name = new_name
        print(f"Snake renamed from '{old_name}' to '{self.name}'.")
```

```python


# Instantiating objects
cobra = Snake("King Cobra", venomous=True)
python_snake = Snake("Burmese Python", venomous=False)

print(f"Snake 1: {cobra.name} | Species: {cobra.species} | Venomous: {cobra.venomous}")
print(f"Snake 2: {python_snake.name} | Species: {python_snake.species} | Venomous: {python_snake.venomous}")

```

```python
# Modifying instance state via method
cobra.change_name("Naja naja")
```

### 2. Building an AI Model Estimator Class
In machine learning, models are structured as classes with hyperparameters stored in `__init__` and behavior implemented in methods like `train()` and `predict()`.


```python
class LinearRegressionEstimator:
    def __init__(self, learning_rate: float = 0.01, epochs: int = 100):
        # Hyperparameters (Instance state)
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weight = 0.0
        self.bias = 0.0
        self.is_fitted = False
        
    def fit(self, X: list, y: list):
        """Simulates training weights on data."""
        print(f"Training on {len(X)} samples for {self.epochs} epochs (LR={self.learning_rate})...")
        # Simplified analytical slope estimation
        self.weight = 2.5
        self.bias = 1.0
        self.is_fitted = True
        print(f"Fit Complete -> Weight: {self.weight}, Bias: {self.bias}")
        return self
        
    def predict(self, X: list) -> list:
        if not self.is_fitted:
            raise RuntimeError("Model must be fitted before predicting!")
        return [self.weight * x + self.bias for x in X]

# Instantiating and executing estimator
model = LinearRegressionEstimator(learning_rate=0.05, epochs=500)
X_train = [1.0, 2.0, 3.0, 4.0]
y_train = [3.5, 6.0, 8.5, 11.0]

model.fit(X_train, y_train)

X_test = [5.0, 10.0]
preds = model.predict(X_test)
print(f"Predictions for {X_test} -> {preds}")
```

### 3. String Representation Dunder Methods: `__str__` and `__repr__`
- `__str__`: User-friendly string representation when calling `print(obj)`.
- `__repr__`: Unambiguous developer/debugging string representation.


```python
class LayerConfig:
    def __init__(self, name: str, units: int, activation: str = "relu"):
        self.name = name
        self.units = units
        self.activation = activation
        
    def __str__(self) -> str:
        return f"Layer '{self.name}' ({self.units} units, activation={self.activation})"
        
    def __repr__(self) -> str:
        return f"LayerConfig(name='{self.name}', units={self.units}, activation='{self.activation}')"

layer = LayerConfig("Dense_1", 256, "gelu")
print("User view (str):", str(layer))
print("Debug view (repr):", repr(layer))
```

### 🏋️ Student Practice Challenge:
Create a class `DatasetSampler` with:
1. `__init__(self, data: list, batch_size: int = 32)`
2. A method `get_total_batches(self) -> int` returning the number of complete batches.
3. A method `sample_batch(self, batch_idx: int) -> list` returning the slice of data for that batch index.


```python
# Write your solution below:
class DatasetSampler:
    def __init__(self, data: list, batch_size: int = 32):
        self.data = data
        self.batch_size = batch_size
        
    def get_total_batches(self) -> int:
        return (len(self.data) + self.batch_size - 1) // self.batch_size
        
    def sample_batch(self, batch_idx: int) -> list:
        start = batch_idx * self.batch_size
        end = start + self.batch_size
        return self.data[start:end]

# Test sampler
sample_dataset = list(range(100))
sampler = DatasetSampler(sample_dataset, batch_size=30)
print(f"Total Batches: {sampler.get_total_batches()}")
print(f"Batch 0: {sampler.sample_batch(0)}")
print(f"Batch 3: {sampler.sample_batch(3)}")
```
