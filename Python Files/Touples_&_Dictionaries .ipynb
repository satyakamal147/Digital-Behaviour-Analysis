### 1. Tuples: Immutability and Packing / Unpacking
Tuples cannot be modified after creation, which prevents accidental mutation of critical model dimensions.


```python
# Defining a tuple for tensor shape
tensor_shape = (32, 3, 224, 224)  # (Batch, Channels, Height, Width)
print(f"Tensor Shape: {tensor_shape} (Type: {type(tensor_shape)})")

# Tuple unpacking into descriptive variable names
batch_size, channels, height, width = tensor_shape
print(f"Unpacked -> Batch: {batch_size}, Channels: {channels}, H: {height}, W: {width}")

# Immutability verification: attempting assignment raises TypeError
try:
    tensor_shape[0] = 64
except TypeError as e:
    print(f"\nExpected Immutability Error: {e}")
```

### 2. Dictionaries: Key-Value Hash Maps & Methods
Dictionaries store data in associative key-value pairs using hash tables for $O(1)$ fast key retrieval.


```python
# Model hyperparameters configuration dictionary
model_config = {
    "model_name": "ResNet-50",
    "learning_rate": 0.0005,
    "batch_size": 32,
    "optimizer": "AdamW",
    "dropout": 0.3
}

print("Model Config Dictionary:\n", model_config)

# Accessing keys safely with .get() (avoids KeyError if key missing)
lr = model_config.get("learning_rate", 0.001)
weight_decay = model_config.get("weight_decay", 1e-4) # Fallback to default
print(f"LR: {lr}, Weight Decay: {weight_decay}")

# Updating dictionary with new parameters
model_config.update({"epochs": 25, "device": "cuda:0"})
print("\nUpdated Config:", model_config)

# Iterating over key-value pairs with .items()
print("\n--- Hyperparameter Summary ---")
for key, value in model_config.items():
    print(f"  {key:<15} : {value}")
```

### 3. Building a Tokenizer Vocabulary with Dictionary Comprehensions
In NLP, dictionary comprehensions create bidirectional word-to-ID and ID-to-word mappings.


```python
# Vocabulary token list
vocabulary = ["<PAD>", "<UNK>", "<SOS>", "<EOS>", "python", "artificial", "intelligence", "neural", "network"]

# Word-to-Index (Encoding map)
word2idx = {token: idx for idx, token in enumerate(vocabulary)}
print("Word to Index Mapping (word2idx):\n", word2idx)

# Index-to-Word (Decoding map)
idx2word = {idx: token for token, idx in word2idx.items()}
print("\nIndex to Word Mapping (idx2word):\n", idx2word)

# Encoding a sample sentence
sentence = "python artificial neural network".split()
encoded_ids = [word2idx.get(w, word2idx["<UNK>"]) for w in sentence]
print("\nEncoded Sequence:", encoded_ids)

# Decoding back to tokens
decoded_sentence = " ".join([idx2word[idx] for idx in encoded_ids])
print("Decoded Sentence:", decoded_sentence)
```

### 4. Sets: Uniqueness and Mathematical Set Operations
Sets automatically eliminate duplicate items and provide high-speed membership testing.


```python
# Eliminating duplicate tokens from raw corpus
corpus_tokens = ["ai", "model", "python", "ai", "data", "python", "neural", "model"]
unique_vocab = set(corpus_tokens)
print("Unique Vocabulary Set:", unique_vocab)
```

```python


# Mathematical Set Operations
set_a = {"python", "c++", "java", "rust"}
set_b = {"python", "julia", "r", "c++"}

print("Union (All languages) [A | B]:", set_a | set_b)
print("Intersection (Common) [A & B]:", set_a & set_b)
print("Difference (In A but not in B) [A - B]:", set_a - set_b)
print("Symmetric Difference (In either but not both) [A ^ B]:", set_a ^ set_b)
```

### 🏋️ Student Practice Challenge:
Given a list of predicted class labels `predictions = ["cat", "dog", "cat", "bird", "dog", "cat", "fish", "dog"]`:
1. Use a dictionary comprehension or loop to count the frequency of each class.
2. Find the set of unique classes present in the predictions.


```python
# Write your solution below:
predictions = ["cat", "dog", "cat", "bird", "dog", "cat", "fish", "dog"]

# 1. Frequency count dictionary
class_counts = {}
for item in predictions:
    class_counts[item] = class_counts.get(item, 0) + 1

# 2. Unique set
unique_classes = set(predictions)

print("Class Frequencies:", class_counts)
print("Unique Classes:", unique_classes)
```
