### 🧠 Theory & Concepts:
- In Python, a variable is **a reference (pointer)** to an object located in heap memory.
- When you write `x = 5`, Python creates an integer object `5` in memory and binds the label `x` to it.
- **`id(object)`:** Returns the unique identity (memory address in CPython) of the object.
- **`type(object)`:** Returns the data type class of the object.


### 1. Assigning Variables & Inspecting Memory References with `id()`
Let's create variables of different types and inspect their memory addresses and data types.


```python
# Numeric variables
item_count = 100          # int
learning_rate = 0.0025    # float
complex_weight = 3 + 4j   # complex

print(f"item_count: value={item_count}, type={type(item_count)}, id={id(item_count)}")
print(f"learning_rate: value={learning_rate}, type={type(learning_rate)}, id={id(learning_rate)}")
print(f"complex_weight: value={complex_weight}, type={type(complex_weight)}, id={id(complex_weight)}")
```

### 2. Memory Reassignment and Variable Re-referencing
When you reassign a variable to a new value, Python creates a new object and points the variable label to the new memory address.


```python
a = 50
print(f"Initial a: value={a}, memory_address={id(a)}")

# Reassigning variable a
a = 150
print(f"Reassigned a: value={a}, memory_address={id(a)} (Notice the address changed!)")
```

### 3. Calculating Total Costs & Real-World Variables Example
Let's see an example calculating shopping or inventory prices dynamically.


```python
# Product price calculation
tshirt = 450.0
jeans = 1200.0
shoes = 2500.0

total_cart = tshirt + jeans + shoes
tax = total_cart * 0.18
final_payable = total_cart + tax

print(f"Subtotal: ₹{total_cart:.2f}")
print(f"GST (18%): ₹{tax:.2f}")
print(f"Grand Total: ₹{final_payable:.2f}")
```

### 4. Numeric Types in Detail: Integers, Floats, and Complex Numbers
Python supports arbitrary precision integers and IEEE-754 standard floating-point numbers.


```python
# Large integers with arbitrary precision
large_int = 10**30
print("Large Integer:", large_int)

# Floating point precision and scientific notation
exp_float = 1.5e-4  # 0.00015
print("Scientific Float:", exp_float)

# Complex numbers (useful in Fourier Transforms and Signal AI)
z = 2 + 5j
print(f"Complex: {z} | Real Part: {z.real} | Imaginary Part: {z.imag} | Conjugate: {z.conjugate()}")
```

### 5. Type Casting / Explicit Type Conversion
In AI pipelines, data read from CSVs or APIs often arrives as strings. Explicit casting converts text to numerical representations.


```python
# Converting string to numeric for ML model input
raw_batch_str = "128"
raw_loss_str = "0.0458"

batch_int = int(raw_batch_str)
loss_float = float(raw_loss_str)

print(f"batch_int: {batch_int} ({type(batch_int)})")
print(f"loss_float: {loss_float} ({type(loss_float)})")

# Converting numbers back to string for logging / formatting
log_entry = "Batch Size: " + str(batch_int) + " | Loss: " + str(loss_float)
print(log_entry)

# Boolean casting rules: 0, None, empty containers -> False; everything else -> True
print("bool(0):", bool(0))
print("bool(1):", bool(1))
print("bool(''):", bool(''))
print("bool('AI'):", bool('AI'))
```
