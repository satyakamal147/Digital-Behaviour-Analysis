### 🧠 Theory & Concepts:
- **`if-elif-else`:** Evaluates conditional expressions in top-down order; executes the block under the first `True` condition and exits the ladder.
- **`for` loop:** Definite iteration over any iterable object (list, tuple, range, string, generator).
- **`while` loop:** Indefinite iteration running as long as a condition remains `True`.
- **`break`:** Immediately terminates the enclosing loop.
- **`continue`:** Skips the rest of the current iteration and jumps directly to the next cycle.
- **`pass`:** A null statement placeholder to satisfy Python indentation syntax.


### 1. Conditional Statements: `if`, `elif`, and `else`
Let's see multi-tier grade classification and loss threshold evaluation.


```python
# Classification of model validation loss
val_loss = 0.28

if val_loss < 0.10:
    status = "Outstanding Convergence"
elif val_loss < 0.30:
    status = "Good Performance - Continue Training"
elif val_loss < 0.70:
    status = "Moderate Loss - Consider Tuning Hyperparameters"
else:
    status = "Underfitting / High Loss - Model Diverging"

print(f"Validation Loss: {val_loss} -> Status: {status}")
```

### 2. The `for` Loop with `range()`
The `range(start, stop, step)` function generates a sequence of numbers lazily.


```python
# Standard training epoch iteration
total_epochs = 5

for epoch in range(1, total_epochs + 1):
    simulated_loss = 1.0 / epoch
    print(f"Epoch {epoch}/{total_epochs} completed | Loss: {simulated_loss:.4f}")
```

### 3. Iterating with `enumerate()` and `zip()`
- `enumerate(iterable, start=0)`: Yields pairs of `(index, item)`, eliminating manual counter variables.
- `zip(iter1, iter2, ...)`: Combines multiple iterables in parallel, yielding tuples of matched elements.


```python
# Using enumerate for feature inspection
feature_names = ["Age", "Income", "Credit_Score", "Loan_Amount"]

print("--- Feature Index Map ---")
for idx, feature in enumerate(feature_names, start=1):
    print(f"Feature #{idx}: {feature}")
```

```python


print("\n--- Parallel Stream with zip() ---")
# Pairing features with their calculated importance weights
importance_weights = [0.12, 0.45, 0.33, 0.10]

for feat, weight in zip(feature_names, importance_weights):
    print(f"{feat:<15} -> Importance: {weight:.2f}")
```

### 4. The `while` Loop & Convergence Check
`while` loops run until the stopping condition evaluates to False. Ideal for optimization algorithms until error drops below a tolerance threshold.


```python
current_loss = 1.50
learning_rate = 0.3
tolerance = 0.05
step = 0

print("--- Gradient Descent Simulation ---")
while current_loss > tolerance:
    step += 1
    # Simulated gradient step: loss decreases by a factor
    current_loss -= learning_rate * current_loss
    print(f"Step {step}: Current Loss = {current_loss:.6f}")

print(f"Converged below tolerance ({tolerance}) in {step} steps!")
```

### 5. Loop Control Statements: `break`, `continue`, and `pass`
- `break`: Aborts loop (e.g. early stopping when validation loss degrades).
- `continue`: Skips remainder of current epoch (e.g. skipping corrupted batch).
- `pass`: Placeholder for future code implementation.


```python
# Loop control demonstration in batch processing
batch_data = [10, 20, -1, 40, 50, 999, 70]

print("--- Batch Processing with Loop Controls ---")
for batch_id, val in enumerate(batch_data):
    if val < 0:
        print(f"Batch {batch_id}: Corrupted data ({val}). Skipping (continue)...")
        continue
    
    if val == 999:
        print(f"Batch {batch_id}: Fatal Poison Value ({val}) encountered. Aborting training (break)!")
        break
        
    print(f"Batch {batch_id}: Successfully processed value = {val}")
```

### 6. The `for...else` Construct in Python
In Python, a `for` loop can have an optional `else` block. It executes **only if the loop completed all iterations without encountering a `break`**.


```python
# Searching for an target token in vocabulary
vocab = ["bert", "gpt", "t5", "roberta", "claude"]
target = "llama"

for token in vocab:
    if token == target:
        print(f"Found target '{target}' in vocabulary!")
        break
else:
    print(f"Target '{target}' was NOT found in vocabulary (completed full search).")
```

Write a loop that iterates through numbers from `1` to `20`. For each number:
- If divisible by 3, print `"Fizz"`
- If divisible by 5, print `"Buzz"`
- If divisible by both 3 and 5, print `"FizzBuzz"`
- Otherwise print the number itself.


```python
# Write your solution below:
for n in range(1, 21):
    if n % 3 == 0 and n % 5 == 0:
        print(f"{n}: FizzBuzz")
    elif n % 3 == 0:
        print(f"{n}: Fizz")
    elif n % 5 == 0:
        print(f"{n}: Buzz")
    else:
        print(f"{n}: {n}")
```
