### 1. Arithmetic Operators
Arithmetic operators perform mathematical calculations:
- `+` (Addition), `-` (Subtraction), `*` (Multiplication)
- `/` (Float division - always returns float)
- `//` (Floor / Integer division - truncates fractional part)
- `%` (Modulo - returns remainder of division)
- `**` (Exponentiation / Power)


```python
a = 17
b = 5

print(f"a = {a}, b = {b}")
print(f"Addition (a + b): {a + b}")
print(f"Subtraction (a - b): {a - b}")
print(f"Multiplication (a * b): {a * b}")
print(f"Float Division (a / b): {a / b}")
print(f"Floor Division (a // b): {a // b}")
print(f"Modulo Remainder (a % b): {a % b}")
print(f"Exponentiation (a ** b): {a ** b}")
```

### 2. Comparison / Relational Operators
Comparison operators evaluate relational expressions and always return a boolean: `True` or `False`.


```python
score_a = 88.5
score_b = 92.0

print(f"score_a == score_b: {score_a == score_b}")  # Equal to
print(f"score_a != score_b: {score_a != score_b}")  # Not equal to
print(f"score_a < score_b:  {score_a < score_b}")   # Less than
print(f"score_a > score_b:  {score_a > score_b}")   # Greater than
print(f"score_a <= 88.5:    {score_a <= 88.5}")     # Less than or equal
print(f"score_b >= 90.0:    {score_b >= 90.0}")     # Greater than or equal
```

### 3. Logical Operators & Short-Circuit Evaluation
Logical operators combine boolean expressions:
- `and`: Returns `True` only if **both** operands are True.
- `or`: Returns `True` if **at least one** operand is True.
- `not`: Inverts the boolean truth value.

> ⚡ **Short-Circuit Evaluation:** For `and`, if the first operand is False, the second is not evaluated. For `or`, if the first operand is True, the second is skipped.


```python
has_gpu = True
epochs_completed = 50
loss_below_threshold = True

# Logical and
should_save_checkpoint = has_gpu and loss_below_threshold
print("Save Checkpoint (and):", should_save_checkpoint)

# Logical or
is_ready_for_eval = (epochs_completed >= 50) or loss_below_threshold
print("Ready for Eval (or):", is_ready_for_eval)

# Logical not
is_training_active = False
print("Is Training Paused (not):", not is_training_active)
```

### 4. Bitwise Operators
Bitwise operators act on the binary bit representations of integers:
- `&` (AND), `|` (OR), `^` (XOR), `~` (NOT / Inversion)
- `<<` (Left shift - multiplies by powers of 2)
- `>>` (Right shift - divides by powers of 2)


```python
x = 10  # Binary: 0000 1010
y = 4   # Binary: 0000 0100

print(f"x = {x} ({bin(x)}), y = {y} ({bin(y)})")
print(f"x & y (Bitwise AND): {x & y} ({bin(x & y)})")
print(f"x | y (Bitwise OR):  {x | y} ({bin(x | y)})")
print(f"x ^ y (Bitwise XOR): {x ^ y} ({bin(x ^ y)})")
print(f"~x (Bitwise NOT):    {~x} ({bin(~x)})")
print(f"x << 2 (Left Shift): {x << 2} ({bin(x << 2)}) [10 * 4 = 40]")
print(f"x >> 1 (Right Shift):{x >> 1} ({bin(x >> 1)}) [10 // 2 = 5]")
```

### 5. Identity Operators (`is` vs `==`)
- `==` compares **value equivalence** (whether contents are identical).
- `is` compares **memory identity** (whether both operands reference the exact same heap memory address).


```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

print("list1 == list2 (Value equality):", list1 == list2) # True
print("list1 is list2 (Object identity):", list1 is list2) # False (different objects)
print("list1 is list3 (Object identity):", list1 is list3) # True (same memory reference)
print(f"id(list1)={id(list1)}, id(list2)={id(list2)}, id(list3)={id(list3)}")
```

### 6. Membership Operators (`in` and `not in`)
Membership operators test if a value is contained within a sequence (string, list, tuple, set, or dictionary).


```python
# Checking token in vocabulary list
vocab = ["transformer", "attention", "token", "embedding", "loss"]

print("'attention' in vocab:", "attention" in vocab)
print("'gradient' in vocab:", "gradient" in vocab)
print("'backprop' not in vocab:", "backprop" not in vocab)

# Checking key presence in dictionary
model_config = {"lr": 0.001, "batch_size": 32, "epochs": 10}
print("'lr' in model_config:", "lr" in model_config)
print("'dropout' in model_config:", "dropout" in model_config)
```
