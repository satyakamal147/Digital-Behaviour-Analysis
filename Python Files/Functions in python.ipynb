### 🧠 Theory & Concepts:
- **Functions:** Named blocks of reusable code that take inputs (arguments), perform operations, and return output via the `return` statement.
- **`*args`:** Gathers excess positional arguments into a tuple.
- **`**kwargs`:** Gathers excess keyword arguments into a dictionary.
- **LEGB Scope:** Python searches for variable names in this priority order:
  1. **L**ocal: Inside the current function.
  2. **E**nclosing: In any outer enclosing functions (nested scopes).
  3. **G**lobal: Module-level variables.
  4. **B**uilt-in: Pre-loaded Python functions (`print`, `len`, `range`).


### 1. Defining Functions & Default Arguments
Functions are declared using `def function_name(params):`. Default values provide fallback arguments.


```python
# Function definition with type annotations and docstring
def compute_loss(y_true: float, y_pred: float, loss_type: str = "MSE") -> float:
    """
    Computes loss between true target and predicted value.
    Supports 'MSE' (Mean Squared Error) and 'MAE' (Mean Absolute Error).
    """
    if loss_type == "MSE":
        return (y_true - y_pred) ** 2
    elif loss_type == "MAE":
        return abs(y_true - y_pred)
    else:
        raise ValueError(f"Unsupported loss_type: {loss_type}")

# Calling the function with positional and default arguments
mse_loss = compute_loss(10.0, 8.5)
mae_loss = compute_loss(10.0, 8.5, loss_type="MAE")

print(f"MSE Loss: {mse_loss:.4f}")
print(f"MAE Loss: {mae_loss:.4f}")
print("Docstring:\n", compute_loss.__doc__)
```

### 3. Variable Scope & The LEGB Rule (`global` vs `nonlocal`)
Variables defined inside a function belong to the local scope unless declared with `global` or `nonlocal`.


```python
# Global variable
global_model_version = "v1.0.0"

def train_module():
    # Local variable
    local_epochs = 100
    
    def inner_step():
        # Enclosing scope access
        print(f"Inside inner_step: Epochs={local_epochs}, Model={global_model_version}")
    
    inner_step()
    print("Inside train_module: Local epochs =", local_epochs)

train_module()
print("In global scope: Model Version =", global_model_version)
```

### 4. Lambda Functions (Anonymous Functions)
Lambda functions are concise, single-expression anonymous functions created with `lambda arguments: expression`.


```python
# Standard function vs Lambda function
def standard_relu(x):
    return max(0.0, x)

# Equivalent Lambda function
lambda_relu = lambda x: max(0.0, x)

print("Standard ReLU(-5.5):", standard_relu(-5.5))
print("Lambda ReLU(-5.5):", lambda_relu(-5.5))
print("Lambda ReLU(3.2):", lambda_relu(3.2))
```

### 5. Functional Tools: `map()`, `filter()`, and `reduce()`
- `map(func, iterable)`: Applies `func` to every item in the sequence.
- `filter(predicate, iterable)`: Keeps only items where `predicate(item)` evaluates to True.
- `reduce(func, iterable)`: Cumulatively applies a 2-argument function across the sequence.


```python
from functools import reduce

raw_activations = [-1.5, 0.0, 2.3, -0.4, 4.1, 1.8]

# 1. map(): Apply ReLU activation across all values
relu_results = list(map(lambda x: max(0.0, x), raw_activations))
print("1. Map (ReLU):", relu_results)
```

```python
# 2. filter(): Retain only active (> 0) neurons
active_neurons = list(filter(lambda x: x > 0.0, relu_results))
print("2. Filter (Active Neurons):", active_neurons)
```

```python



# 3. reduce(): Calculate sum of active activations
total_signal = reduce(lambda acc, val: acc + val, active_neurons)
print(f"3. Reduce (Total Signal Sum): {total_signal:.2f}")
```

### 🏋️ Student Practice Challenge:
Write a function `softmax_approximation(logits: list) -> list` that:
1. Takes a list of numerical logits.
2. Computes $e^{x}$ for each logit using `map()` and `math.exp`.
3. Computes the total sum of exponentials using `reduce()` or `sum()`.
4. Returns the normalized probability distribution.


```python
import math

# Write your solution below:
def softmax_approximation(logits: list) -> list:
    exp_vals = list(map(lambda x: math.exp(x), logits))
    total_exp = sum(exp_vals)
    probabilities = [round(v / total_exp, 4) for v in exp_vals]
    return probabilities

test_logits = [2.0, 1.0, 0.1]
probs = softmax_approximation(test_logits)
print(f"Logits: {test_logits} -> Softmax Probabilities: {probs} (Sum={sum(probs):.2f})")
```
