### 1. Pillar 1: Encapsulation & Data Hiding
Encapsulation binds data attributes with their methods and restricts direct access to internal variables:
- `public_var`: Accessible from anywhere.
- `_protected_var`: Internal by convention.
- `__private_var`: Name-mangled by Python (accessed as `_ClassName__private_var`).
- `@property`: Clean getter and setter interface with validation.


```python
class ModelCheckpoint:
    def __init__(self, model_name: str, initial_lr: float = 0.001):
        self.model_name = model_name       # Public attribute
        self._epoch = 0                    # Protected attribute (convention)
        self.__learning_rate = initial_lr  # Private attribute (name-mangled)
        
    @property
    def learning_rate(self) -> float:
        """Getter method for private learning rate."""
        return self.__learning_rate
        
    @learning_rate.setter
    def learning_rate(self, new_lr: float):
        """Setter with input validation to prevent invalid learning rates."""
        if new_lr <= 0:
            raise ValueError(f"Learning rate must be positive! Received: {new_lr}")
        print(f"Updating learning rate from {self.__learning_rate} to {new_lr}")
        self.__learning_rate = new_lr

# Testing Encapsulation
ckpt = ModelCheckpoint("BERT-Base", initial_lr=0.0001)
print(f"Current LR via property: {ckpt.learning_rate}")

# Valid update
ckpt.learning_rate = 0.0005

# Invalid update catching exception
try:
    ckpt.learning_rate = -0.01
except ValueError as e:
    print(f"[Caught Invalid Update] {e}")
```

### 2. Pillar 2: Inheritance & `super()`
Inheritance allows a child class to inherit attributes and methods from a parent class, promoting code reusability.


```python
# Parent Base Class
class BaseNeuralLayer:
    def __init__(self, layer_name: str):
        self.layer_name = layer_name
        self.trainable = True
        
    def summary(self):
        print(f"Layer: {self.layer_name} | Trainable: {self.trainable}")

# Child Class inheriting BaseNeuralLayer
class DenseLayer(BaseNeuralLayer):
    def __init__(self, layer_name: str, in_features: int, out_features: int):
        # Call parent constructor using super()
        super().__init__(layer_name=layer_name)
        self.in_features = in_features
        self.out_features = out_features
        
    def summary(self):
        # Overriding parent method while calling base info
        super().summary()
        print(f"  Shape: ({self.in_features} -> {self.out_features}) | Total Weights: {self.in_features * self.out_features}")

dense = DenseLayer("fc1", in_features=784, out_features=128)
dense.summary()
```

### 3. Pillar 3: Polymorphism
Polymorphism allows different classes to implement methods with the identical name and signature, enabling dynamic execution.


```python
class DecisionTreeModel:
    def predict(self, x: float) -> str:
        return "Class A" if x > 5.0 else "Class B"

class SupportVectorModel:
    def predict(self, x: float) -> str:
        return "Class A" if x > 4.2 else "Class B"

class NeuralNetModel:
    def predict(self, x: float) -> str:
        return "Class A" if x > 4.8 else "Class B"

# Polymorphic Model Evaluator Function
def evaluate_inference_pipeline(models: list, test_val: float):
    print(f"--- Running Inference on Input Value: {test_val} ---")
    for model in models:
        # Uniform method call .predict() across completely different architectures
        prediction = model.predict(test_val)
        print(f"{model.__class__.__name__:<20} Prediction: {prediction}")

ensemble = [DecisionTreeModel(), SupportVectorModel(), NeuralNetModel()]
evaluate_inference_pipeline(ensemble, test_val=4.5)
```

### 4. Pillar 4: Abstraction with Abstract Base Classes (`abc.ABC`)
Abstraction hides complex implementation details and exposes only essential interfaces via `@abstractmethod`.


```python
from abc import ABC, abstractmethod
import math

class BaseLossFunction(ABC):
    """Abstract interface enforcing loss calculation contracts."""
    @abstractmethod
    def calculate_loss(self, y_true: list, y_pred: list) -> float:
        pass

# Concrete Implementation 1: MSE Loss
class MSELoss(BaseLossFunction):
    def calculate_loss(self, y_true: list, y_pred: list) -> float:
        return sum((yt - yp) ** 2 for yt, yp in zip(y_true, y_pred)) / len(y_true)

# Concrete Implementation 2: MAE Loss
class MAELoss(BaseLossFunction):
    def calculate_loss(self, y_true: list, y_pred: list) -> float:
        return sum(abs(yt - yp) for yt, yp in zip(y_true, y_pred)) / len(y_true)

y_actual = [10.0, 20.0, 30.0]
y_forecast = [11.0, 19.0, 28.0]

mse_calc = MSELoss()
mae_calc = MAELoss()

print(f"MSE Loss: {mse_calc.calculate_loss(y_actual, y_forecast):.4f}")
print(f"MAE Loss: {mae_calc.calculate_loss(y_actual, y_forecast):.4f}")
```

### 5. Dunder (Magic) Methods for Custom Datasets
Implementing `__len__`, `__getitem__`, and `__call__` allows custom classes to behave like native Python sequences and callable functions (identical to PyTorch `Dataset` and `nn.Module`).


```python
class CustomAIDataset:
    def __init__(self, data_list: list):
        self._data = data_list
        
    def __len__(self) -> int:
        return len(self._data)
        
    def __getitem__(self, idx: int):
        return self._data[idx]
        
    def __call__(self, scale_factor: float):
        """Calling dataset instance scales all features."""
        return [x * scale_factor for x in self._data]

dataset = CustomAIDataset([10.0, 20.0, 30.0, 40.0, 50.0])
print(f"Dataset Length (len(dataset)): {len(dataset)}")
print(f"Item at Index 2 (dataset[2]): {dataset[2]}")
print(f"Scaled Dataset (dataset(0.1)): {dataset(0.1)}")
```

### 🏋️ Student Practice Challenge:
Create an Abstract Base Class `BasePreprocessor(ABC)` with an abstract method `transform(self, data: list) -> list`. Then create a derived class `MinMaxNormalizer` that implements `transform()` by scaling values to $[0.0, 1.0]$ using $\frac{x - \min}{\max - \min}$.


```python
from abc import ABC, abstractmethod

# Write your solution below:
class BasePreprocessor(ABC):
    @abstractmethod
    def transform(self, data: list) -> list:
        pass

class MinMaxNormalizer(BasePreprocessor):
    def transform(self, data: list) -> list:
        min_val = min(data)
        max_val = max(data)
        diff = max_val - min_val if max_val != min_val else 1.0
        return [round((x - min_val) / diff, 4) for x in data]

normalizer = MinMaxNormalizer()
raw_scores = [100, 250, 400, 550, 1000]
print("Raw Scores:", raw_scores)
print("Normalized Scores [0.0 - 1.0]:", normalizer.transform(raw_scores))
```
