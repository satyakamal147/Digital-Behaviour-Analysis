### 🧠 Theory & Concepts:
- **Interpreted Nature:** Python code is compiled into bytecode (`.pyc`) which is executed line-by-line by the Python Virtual Machine (PVM).
- **Indentation as Syntax:** Unlike C/C++/Java which use `{}` curly braces to define blocks, Python strictly enforces indentation (4 spaces) to define code blocks.
- **Dynamic Typing:** You do not declare types explicitly (e.g. `int x = 5;`). Python dynamically binds the type at runtime based on the value assigned.


### 1. Basic Output & The `print()` Function
The `print()` function is Python's standard output mechanism. It prints strings, variables, or expressions to the console with an optional separator (`sep`) and end character (`end`).


```python
# Basic print statement
print("Hello, Welcome to GRIET AIML Program!")

# Printing multiple arguments with custom separator
print("Python", "Programming", "for", "AI", sep=" -> ")

# Custom end parameter (defaults to newline '\n')
print("Training in progress...", end=" [DONE]\n")
```

```python
print(""" Aerial video shows the immense destruction in villages in Nepal along the Trishuli river, following a catastrophic glacial flood on Wednesday. Authorities are still searching for around 2,500 missing people.""")
```

### 2. Lines & Indentation
In Python, all statements with the same distance to the right belong to the same block of code.
> ⚠️ **Key Rule:** An `IndentationError` occurs if tabs and spaces are mixed or if indentation levels are inconsistent.


```python
# Correct indentation in conditional statements
is_gpu_available = True

if is_gpu_available:
    print("Device: CUDA GPU Enabled")
    print("Allocating tensor memory...")
else:
    print("Device: CPU Mode")
    print("Allocating host RAM...")
```

### 3. Multi-line Statements with Continuation Characters
Long statements can span multiple lines using the backslash continuation character `\` or implicitly within parentheses `()`, brackets `[]`, and braces `{}`.


```python
# Explicit line continuation with backslash \
total_dataset_records = 15000 + \
                        25000 + \
                        35000
print("Total Dataset Records:", total_dataset_records)

# Implicit line continuation inside brackets/parentheses (Recommended)
model_layers = [
    "InputLayer(shape=(28, 28))",
    "DenseLayer(units=128, activation='relu')",
    "DropoutLayer(rate=0.2)",
    "OutputLayer(units=10, activation='softmax')"
]
print("Pipeline Architecture:", model_layers)
```

### 4. Multiple Assignments in a Single Line
Python allows initializing multiple variables simultaneously using tuple unpacking.


```python
# Multiple assignment: Same value
epoch_step = batch_counter = gradient_step = 0
print(f"Step: {epoch_step}, Batch: {batch_counter}, Grad: {gradient_step}")

# Multiple assignment: Different values (Unpacking)
learning_rate, batch_size, model_name = 0.001, 64, "Transformer-Encoder"
print(f"Model: {model_name} | LR: {learning_rate} | Batch: {batch_size}")
```

### 5. Comments and Documentation
Comments are non-executable text lines ignored by the interpreter:
- **Single-line comments:** Start with `#`
- **Multi-line docstrings:** Enclosed in triple quotes `'''` or `"""` and used for function/module documentation.


```python
# This is a single-line comment explaining feature scaling
feature_x = 42.0  # Feature value

"""
Multi-line Docstring:
This block can be used for descriptive explanations, 
mathematical derivations, or module-level docstrings.
"""
print("Feature value initialized to:", feature_x)
```
